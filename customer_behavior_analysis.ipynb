```python
import pandas as pd
```

```python
path = "customer_behaviour.csv"
df = pd.read_csv(path, sep=';', decimal=',')
df.head()
```

```python
df.isnull().sum()
```

```python
df.duplicated()
```

```python
dick_new_names = {}
for column in df.columns:
    dick_new_names[column] = str(column).lower().strip().replace(" ","_")
print(dick_new_names)

df.rename(columns=dick_new_names,inplace=True)
```

```python
print(df.shape)
print(df.info())
```

```python
df.describe()
```

```python
df.describe(exclude=['float', 'int'])
```

```python
list_of_columns = df.columns.tolist()

print(len(list_of_columns))
list_of_columns.remove('credit_card_info_save')
list_of_columns.remove('push_status')
list_of_columns.remove('user_id')
list_of_columns.remove('location_code')
list_of_columns.remove('churn')
print(len(list_of_columns))
print(list_of_columns)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
fig , axes = plt.subplots(5, 3, figsize=(24, 24))
fig.suptitle("Distribution of Features", size= 48)

for i in range(5):
    for j in range(3):
        sns.histplot(df[list_of_columns[3*i + j]], ax=axes[i, j])

plt.show()
```

```python
df.churn.value_counts()
```

```python
fig , ax = plt.subplots(1,2,figsize=(10,6))

plt.pie(df['churn'].value_counts(), labels=['Not Churn', 'Churn'], autopct='%0.0f%%', colors=['skyblue', 'salmon'])
cp = sns.countplot(x=df['churn'], ax=ax[0], palette='husl', hue=df['churn'] , legend=True)
cp.set_xlabel('Churn')
cp.set_ylabel('Count')
cp.set_title('Churn Distribution')

for p in cp.patches:
    cp.annotate(format(p.get_height(), '.0f'),
                (p.get_x() + p.get_width() / 2., p.get_height()),
                ha = 'center', va = 'center',
                xytext = (0, 9),
                textcoords = 'offset points')

plt.show()
```

```python
categorical_columns = list(df.select_dtypes(include=['str']).columns)
categorical_columns.append('location_code')
print(categorical_columns)
```

```python
for i in df.select_dtypes(include=['object', 'str']).columns.values:
    df[i] = df[i].replace(['Yes', 'No'], [1, 0])
```

```python
fig, ax = plt.subplots(1,3, figsize=(15,5))
for i, a in zip(categorical_columns, ax):
    temp = df[i].value_counts()
    a.set_title(i)
    sns.barplot(x=temp.index, y=temp.values, ax=a, palette='viridis', hue=temp.index, legend=True)
```
